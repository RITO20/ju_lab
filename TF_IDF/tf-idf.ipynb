```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from typing import List
import re
from collections import Counter
```

```python
data = pd.read_csv("/home/rito/ju_lab/spam.csv",encoding="latin1")
df = pd.DataFrame(data)
```

```python
def tokenize(text: str) -> List[str]:
    # Remove punctuation using regex, keeping words and numbers
    cleaned_text = re.sub(r"[^\w\s]", "", text)
    # Split the cleaned text into words
    tokens = cleaned_text.lower().split()
    return tokens
```

```python
vocal_class = set()

for i in range(1, len(df['v2'])):
    d1 = tokenize(df.loc[i,'v2'])
    d2 = tokenize(df.loc[i-1,'v2'])

    vocal_class = vocal_class.union(d1,d2)
```

```python
def compute_tf(tokens):
    if not tokens:
        return {}
    tf = Counter(tokens)
    total = len(tokens)
    return {word: count / total for word, count in tf.items()}
```

```python
df_count = dict.fromkeys(vocal_class, 0)

for text in df['v2']:
    tokens = set(tokenize(text))  # IMPORTANT: set()
    for word in tokens:
        df_count[word] += 1
```

```python
df_count
```

```python
N = len(df['v2'])

idf = {}
for word, df_val in df_count.items():
    idf[word] = np.log((N + 1) / (df_val + 1)) + 1
```

```python
def compute_tfidf(tf, idf):
    return {word: tf[word] * idf.get(word, 0) for word in tf}
```

```python
tf_idf = []

def compute_text_tfidf(text):
    tokens = tokenize(text)
    tf = compute_tf(tokens)
    return compute_tfidf(tf, idf)
```

```python
df['text_tf_idf'] = df['v2'].apply(compute_text_tfidf)
```
